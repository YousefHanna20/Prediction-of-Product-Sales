```python
import pandas as pd
df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/weight-height male-female - weight-height.csv")
df.head()
```

```python
female_height = df.loc[df['Gender'] == 'Female', 'Height']
female_height.head()
```

```python
mean = female_height.mean()
std = female_height.std()
print(mean)
print(std)
```

```python
from scipy.stats import norm
```

#Question1: Probability of females height between 55.0 and 56.0 .

```python
prob_1 = norm.cdf(56.0, loc = mean, scale=std) - norm.cdf(55.0, loc= mean, scale=std)
print(prob_1)
```

#Question2: Probability of females height less than 5 feet(60 inch).

```python
prob_2 = norm.cdf(60.0, loc=mean, scale=std)
print(prob_2)
```

#Question3: Probability of females height greater than the mean.

```python
prob_3 = norm.cdf(mean, loc=mean, scale=std)
print(prob_3)
```

###This one can actually be answered without calculation. A normal distribution is symmetric around its mean, so exactly 50% of the distribution is above the mean and 50% below it.
